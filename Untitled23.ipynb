```python
# prompt: this above for dtype convert we are doing on train same we have to it on test  which dont have y target in it so make function of above and call on both

import pandas as pd
def convert_dataframe_dtypes(df, data_dict):
    """
    Converts data types of a DataFrame based on a data dictionary and manual overrides.

    Args:
        df (pd.DataFrame): The DataFrame to convert.
        data_dict (pd.DataFrame): DataFrame containing column names and their types.

    Returns:
        pd.DataFrame: The DataFrame with converted data types.
    """
    # --- Step 2: Clean Column Names and Types ---
    data_dict_cleaned = data_dict.copy() # Create a copy to avoid modifying the original
    data_dict_cleaned['masked_column'] = data_dict_cleaned['masked_column'].str.strip()
    data_dict_cleaned['Type'] = data_dict_cleaned['Type'].str.strip().str.lower()

    # --- Step 3: Create column type mappings ---
    numerical_cols = data_dict_cleaned[data_dict_cleaned['Type'] == 'numerical']['masked_column'].tolist()
    categorical_cols = data_dict_cleaned[data_dict_cleaned['Type'] == 'categorical']['masked_column'].tolist()
    one_hot_cols = data_dict_cleaned[data_dict_cleaned['Type'] == 'one hot encoded']['masked_column'].tolist()
    key_cols = data_dict_cleaned[data_dict_cleaned['Type'] == 'key']['masked_column'].tolist()

    # Include your manual overrides
    column_dtypes = {}

    # Apply from data dict
    for col in numerical_cols:
        column_dtypes[col] = 'float64'

    for col in categorical_cols:
        column_dtypes[col] = 'category'

    for col in one_hot_cols:
        column_dtypes[col] = 'int8'

    for col in key_cols:
        column_dtypes[col] = 'string'

    # Manual override: id2, id3 should be Int64, id4, id5 should be datetime
    column_dtypes['id2'] = 'Int64'
    column_dtypes['id3'] = 'Int64'
    column_dtypes['id4'] = 'datetime64[ns]'
    column_dtypes['id5'] = 'datetime64[ns]'


    # --- Step 4: Apply Conversions ---
    for col, dtype in column_dtypes.items():
        if col in df.columns:
            try:
                if dtype == 'category':
                    df[col] = df[col].astype('category')
                elif dtype == 'float64':
                    df[col] = pd.to_numeric(df[col], errors='coerce')
                elif dtype == 'int8':
                    df[col] = pd.to_numeric(df[col], errors='coerce').astype('Int8')
                elif dtype == 'Int64':
                    df[col] = pd.to_numeric(df[col], errors='coerce').astype('Int64')
                elif dtype == 'string':
                    df[col] = df[col].astype('string')
                elif dtype == 'datetime64[ns]':
                     df[col] = pd.to_datetime(df[col], errors='coerce')
            except Exception as e:
                print(f"Failed to convert column {col} to {dtype}: {e}")
    return df

# --- Apply the function to train_df and test_df ---
print("Converting data types for train_df...")
train_df = convert_dataframe_dtypes(train_df, data_dictionary)
print("\ntrain_df info after conversion:")
train_df.info()

print("\nConverting data types for test_df...")
test_df = convert_dataframe_dtypes(test_df, data_dictionary)
print("\ntest_df info after conversion:")
test_df.info()
```

```python
import pandas as pd

def compute_numeric_medians(df):
    """
    Compute medians for numeric columns in the given DataFrame.
    """
    numeric_cols = df.select_dtypes(include=['number']).columns
    return {col: df[col].median() for col in numeric_cols}

def fill_missing_values(df, numeric_medians=None):
    """
    Fills missing values in a DataFrame.
    - Numeric columns: use provided medians if given, else compute from df.
    - Categorical/String: fill with 'missing'.

    Args:
        df (pd.DataFrame): Input DataFrame.
        numeric_medians (dict, optional): Precomputed medians to use for numeric columns.

    Returns:
        pd.DataFrame: DataFrame with missing values filled.
    """
    if df is None:
        print("Input DataFrame is None. Cannot fill missing values.")
        return None

    print("Filling missing values...")

    for col in df.columns:
        if df[col].isnull().any():
            if pd.api.types.is_numeric_dtype(df[col]):
                median_val = numeric_medians.get(col, df[col].median()) if numeric_medians else df[col].median()
                df[col].fillna(median_val, inplace=True)
                print(f"Filled missing values in numeric column '{col}' with median ({median_val}).")
            elif pd.api.types.is_categorical_dtype(df[col]) or pd.api.types.is_object_dtype(df[col]) or pd.api.types.is_string_dtype(df[col]):
                if pd.api.types.is_categorical_dtype(df[col]):
                    if 'missing' not in df[col].cat.categories:
                        df[col] = df[col].cat.add_categories('missing')
                df[col].fillna('missing', inplace=True)
                print(f"Filled missing values in categorical/string column '{col}' with 'missing'.")
            else:
                print(f"Warning: Column '{col}' dtype {df[col].dtype} not explicitly handled.")

    print("Missing value filling complete.\n")
    return df

# === ✅ Correct usage ===

# 1️⃣ Compute train medians once
train_medians = compute_numeric_medians(train_df)

# 2️⃣ Fill train_df using its own medians
train_df = fill_missing_values(train_df, train_medians)

# 3️⃣ Fill test_df using train medians
test_df = fill_missing_values(test_df, train_medians)

# 4️⃣ Fill transactions_df and ad_event_df using their own medians
transactions_df = fill_missing_values(transactions_df)
ad_event_df = fill_missing_values(ad_event_df)



```

```python
# prompt: we forgot about filling datetime cols type which has missing fill it with approriate logic

import pandas as pd
def fill_missing_datetime(df):
    """
    Fills missing values in datetime columns.
    Missing datetime values are filled with a placeholder datetime, e.g., the minimum datetime in the column
    or a specific sentinel value like the Unix epoch start.

    Args:
        df (pd.DataFrame): The input DataFrame.

    Returns:
        pd.DataFrame: The DataFrame with missing datetime values filled.
    """
    if df is None:
        print("Input DataFrame is None. Cannot fill missing datetime values.")
        return None

    print("Filling missing values in datetime columns...")

    for col in df.columns:
        if df[col].isnull().any() and pd.api.types.is_datetime64_any_dtype(df[col]):
            # Option 1: Fill with a specific sentinel datetime (e.g., Unix epoch)
            # fill_value = pd.to_datetime(0, unit='s') # Example: Unix epoch start
            # Option 2: Fill with the minimum existing datetime in the column
            fill_value = df[col].min()

            if pd.isna(fill_value):
                # If the entire column is NaT, fill with a default value
                fill_value = pd.to_datetime('1970-01-01') # Default if column is all NaT

            df[col].fillna(fill_value, inplace=True)
            print(f"Filled missing values in datetime column '{col}' with {fill_value}.")

    print("Missing datetime value filling complete.")
    return df

# Apply the function to your dataframes that might have datetime columns
train_df = fill_missing_datetime(train_df)
ad_event_df = fill_missing_datetime(ad_event_df)
transactions_df = fill_missing_datetime(transactions_df)
test_df = fill_missing_datetime(test_df)

```

```python

def create_features(df, ad_event_df, transactions_df):
    """Creates new features from datetime and historical data."""
    print(f"\nStarting feature engineering for {'train' if 'y' in df.columns else 'test'} data...")

    # === 1. Load + prep ===
    print("Step 1/10: Load & prep data...")
    try:
        # Ensure datetime columns are in datetime format
        df['impression_timestamp'] = pd.to_datetime(df['id4'], errors='coerce')
        df['offer_start_timestamp'] = pd.to_datetime(df['Start Timestamp'], errors='coerce')
        df['offer_end_timestamp'] = pd.to_datetime(df['End Timestamp'], errors='coerce')

        # Create datetime columns for historical data that will be used later
        ad_event_df['impression_timestamp_hist'] = pd.to_datetime(ad_event_df['Impression Timestamp'], errors='coerce')
        ad_event_df['click_timestamp_hist'] = pd.to_datetime(ad_event_df['Click Timestamp'], errors='coerce')
        ad_event_df['is_click_hist'] = ~ad_event_df['click_timestamp_hist'].isna()
        transactions_df['transaction_timestamp'] = pd.to_datetime(transactions_df['Transaction Date'], errors='coerce')

    except KeyError as e:
        print(f"Error: Missing expected column in input DataFrames: {e}. Cannot create features.")
        return df
    except Exception as e:
        print(f"An unexpected error occurred during data prep: {e}. Cannot create features.")
        return df

    placeholder_date = pd.to_datetime('1900-01-01')

    # Fill NaT values in datetime columns with the placeholder date
    for col in ['impression_timestamp', 'offer_start_timestamp', 'offer_end_timestamp']:
        df[col] = df[col].fillna(placeholder_date)
    for col in ['impression_timestamp_hist', 'click_timestamp_hist']:
        ad_event_df[col] = ad_event_df[col].fillna(placeholder_date)
    transactions_df['transaction_timestamp'] = transactions_df['transaction_timestamp'].fillna(placeholder_date)


    # === 2. Offer Popularity ===
    print("Step 2/10: Offer popularity (global CTR)...")
    if 'Offerings ID' in ad_event_df.columns:
        offer_pop = ad_event_df.groupby('Offerings ID').agg(
            offer_total_clicks=('is_click_hist', 'sum'),
            offer_total_impressions=('Customer ID', 'count')
        ).reset_index()
        offer_pop['offer_global_ctr'] = offer_pop['offer_total_clicks'] / (offer_pop['offer_total_impressions'] + 1e-6)
        if 'Offerings ID' in df.columns:
            df = df.merge(offer_pop[['Offerings ID', 'offer_global_ctr']], on='Offerings ID', how='left')
            df['offer_global_ctr'].fillna(offer_pop['offer_global_ctr'].mean(), inplace=True)
        else:
            print("Warning: 'Offerings ID' not found in main DataFrame for merging offer popularity.")
            df['offer_global_ctr'] = offer_pop['offer_global_ctr'].mean()
    else:
        print("Warning: 'Offerings ID' not found in ad_event_df. Skipping offer popularity feature.")
        df['offer_global_ctr'] = 0


    # === 3. User CTR ===
    print("Step 3/10: User CTR baseline...")
    if 'Customer ID' in ad_event_df.columns:
        user_ctr = ad_event_df.groupby('Customer ID').agg(
            user_total_clicks=('is_click_hist', 'sum'),
            user_total_impressions=('Customer ID', 'count')
        ).reset_index()
        user_ctr['user_global_ctr'] = user_ctr['user_total_clicks'] / (user_ctr['user_total_impressions'] + 1e-6)
        if 'Customer ID' in df.columns:
            df = df.merge(user_ctr[['Customer ID', 'user_global_ctr']], on='Customer ID', how='left')
            df['user_global_ctr'].fillna(user_ctr['user_global_ctr'].mean(), inplace=True)
        else:
            print("Warning: 'Customer ID' not found in main DataFrame for merging user CTR.")
            df['user_global_ctr'] = user_ctr['user_global_ctr'].mean()
    else:
        print("Warning: 'Customer ID' not found in ad_event_df. Skipping user CTR feature.")
        df['user_global_ctr'] = 0


    # === 4. Customer–Offer Pair ===
    print("Step 4/10: Pair-specific exposure...")
    if 'Customer ID' in ad_event_df.columns and 'Offerings ID' in ad_event_df.columns:
        pair = ad_event_df.groupby(['Customer ID', 'Offerings ID']).agg(
            pair_total_clicks=('is_click_hist', 'sum'),
            pair_total_impressions=('Customer ID', 'count')
        ).reset_index()
        if 'Customer ID' in df.columns and 'Offerings ID' in df.columns:
            df = df.merge(pair, on=['Customer ID', 'Offerings ID'], how='left')
            df['pair_total_clicks'].fillna(0, inplace=True)
            df['pair_total_impressions'].fillna(0, inplace=True)
        else:
            print("Warning: 'Customer ID' or 'Offerings ID' not found in main DataFrame for merging pair features.")
            df['pair_total_clicks'] = 0
            df['pair_total_impressions'] = 0
    else:
        print("Warning: 'Customer ID' or 'Offerings ID' not found in ad_event_df. Skipping pair features.")
        df['pair_total_clicks'] = 0
        df['pair_total_impressions'] = 0


    # === 5. Offer Validity ===
    print("Step 5/10: Offer age & expiry...")
    df['days_since_launch'] = (df['impression_timestamp'] - df['offer_start_timestamp']).dt.days.clip(lower=0)
    df['days_until_expiry'] = (df['offer_end_timestamp'] - df['impression_timestamp']).dt.days.clip(lower=0)
    df['offer_duration_days'] = (df['offer_end_timestamp'] - df['offer_start_timestamp']).dt.days.clip(lower=0)


    # === 6. Event Time Parts ===
    print("Step 6/10: Hour, day-of-week...")
    df['event_hour'] = df['impression_timestamp'].dt.hour
    df['event_day_of_week'] = df['impression_timestamp'].dt.dayofweek
    df['is_weekend'] = (df['event_day_of_week'] >= 5).astype(int)


    # === 7. Ad Event Rolling Windows ===
    print("Step 7/10: Ad event rolling counts...")
    if 'Customer ID' in df.columns and 'impression_timestamp' in df.columns:
        main_events = df[['Customer ID', 'impression_timestamp']].copy()
        main_events.rename(columns={'impression_timestamp': 'event_time'}, inplace=True)

        if 'Customer ID' in ad_event_df.columns and 'impression_timestamp_hist' in ad_event_df.columns:
            # Use columns that were processed in Step 1
            ad_event_hist = ad_event_df[['Customer ID', 'Offerings ID', 'impression_timestamp_hist']].copy()
            ad_event_hist = ad_event_hist.merge(main_events, on='Customer ID', how='inner').reset_index(drop=True)

            # Filter out historical events that happened after the current event
            ad_event_hist = ad_event_hist[
                ad_event_hist['impression_timestamp_hist'] < ad_event_hist['event_time']
            ].copy()

            print(f"Rows in ad_event_hist after filtering: {len(ad_event_hist)}")

            # FIX: Calculate time difference directly in the DataFrame.
            # This is more robust and avoids the 'ValueError: Operands are not aligned' error.
            ad_event_hist['time_diff'] = ad_event_hist['event_time'] - ad_event_hist['impression_timestamp_hist']

            for window in [7, 30, 90]:
                # Filter for events within the rolling window
                ad_window = ad_event_hist[ad_event_hist['time_diff'] <= pd.Timedelta(days=window)]

                agg = ad_window.groupby('Customer ID').agg(
                    **{
                        f'ad_event_count_last_{window}d': ('Offerings ID', 'count'),
                        f'ad_offer_nunique_last_{window}d': ('Offerings ID', 'nunique')
                    }
                ).reset_index()
                df = df.merge(agg, on='Customer ID', how='left')
                for col in [f'ad_event_count_last_{window}d', f'ad_offer_nunique_last_{window}d']:
                    df[col] = df[col].fillna(0)
        else:
            print("Warning: Required columns not found in ad_event_df. Skipping ad event rolling features.")
            for window in [7, 30, 90]:
                df[f'ad_event_count_last_{window}d'] = 0
                df[f'ad_offer_nunique_last_{window}d'] = 0
    else:
        print("Warning: Required columns not found in main DataFrame. Skipping ad event rolling features.")
        for window in [7, 30, 90]:
            df[f'ad_event_count_last_{window}d'] = 0
            df[f'ad_offer_nunique_last_{window}d'] = 0


    # === 8. Transaction Rolling ===
    print("Step 8/10: Transaction windows & RFM...")
    if 'Customer ID' in transactions_df.columns and 'transaction_timestamp' in transactions_df.columns:
        # FIX: Ensure main_events is defined if it wasn't in Step 7
        if 'main_events' not in locals():
            main_events = df[['Customer ID', 'impression_timestamp']].copy()
            main_events.rename(columns={'impression_timestamp': 'event_time'}, inplace=True)

        # FIX: Use the 'transaction_timestamp' column created in Step 1, not 'Transaction Date'
        trans_hist = transactions_df[['Customer ID', 'Transaction Amount', 'transaction_timestamp']].copy()
        trans_hist = trans_hist.merge(main_events, on='Customer ID', how='inner').reset_index(drop=True)

        # Filter out historical transactions that happened after the current event
        trans_hist = trans_hist[
            trans_hist['transaction_timestamp'] < trans_hist['event_time']
        ].copy()

        print(f"Rows in trans_hist after filtering: {len(trans_hist)}")

        # FIX: Calculate time difference directly in the DataFrame to avoid alignment errors
        trans_hist['time_diff'] = trans_hist['event_time'] - trans_hist['transaction_timestamp']

        for window in [7, 30, 90]:
            # Filter for transactions within the rolling window
            trans_window = trans_hist[trans_hist['time_diff'] <= pd.Timedelta(days=window)]

            agg = trans_window.groupby('Customer ID').agg(
                **{
                    f'trans_count_last_{window}d': ('Transaction Amount', 'count'),
                    f'trans_sum_last_{window}d': ('Transaction Amount', 'sum'),
                    f'trans_mean_last_{window}d': ('Transaction Amount', 'mean')
                }
            ).reset_index()
            df = df.merge(agg, on='Customer ID', how='left')
            for col in [f'trans_count_last_{window}d', f'trans_sum_last_{window}d', f'trans_mean_last_{window}d']:
                df[col] = df[col].fillna(0)
    else:
        print("Warning: Required columns not found in transactions_df. Skipping transaction rolling features.")
        for window in [7, 30, 90]:
            df[f'trans_count_last_{window}d'] = 0
            df[f'trans_sum_last_{window}d'] = 0
            df[f'trans_mean_last_{window}d'] = 0


    # === 9. Transaction Recency ===
    print("Step 9/10: Transaction recency & cadence...")
    # FIX: This section re-calculates trans_hist, so the same fixes from Step 8 must be applied here.
    if 'Customer ID' in transactions_df.columns and 'transaction_timestamp' in transactions_df.columns and 'Customer ID' in df.columns and 'impression_timestamp' in df.columns:
        if 'main_events' not in locals():
            main_events = df[['Customer ID', 'impression_timestamp']].copy()
            main_events.rename(columns={'impression_timestamp': 'event_time'}, inplace=True)

        # FIX: Use the 'transaction_timestamp' column
        trans_hist_recency = transactions_df[['Customer ID', 'transaction_timestamp']].copy()
        trans_hist_recency = trans_hist_recency.merge(main_events, on='Customer ID', how='inner').reset_index(drop=True)
        trans_hist_recency = trans_hist_recency[
            trans_hist_recency['transaction_timestamp'] < trans_hist_recency['event_time']
        ].copy()

        if not trans_hist_recency.empty:
            # FIX: Group by the correct 'transaction_timestamp' column
            user_trans = trans_hist_recency.groupby('Customer ID')['transaction_timestamp'].agg(
                ['min', 'max', 'count', 'nunique']
            ).rename(columns={
                'min': 'first_trans_date',
                'max': 'last_trans_date',
                'count': 'total_trans_count',
                'nunique': 'unique_trans_days'
            }).reset_index()

            df = df.merge(user_trans, on='Customer ID', how='left')

    # Fill missing values for users with no transaction history
    df['first_trans_date'] = df.get('first_trans_date', pd.Series(index=df.index)).fillna(placeholder_date)
    df['last_trans_date'] = df.get('last_trans_date', pd.Series(index=df.index)).fillna(placeholder_date)
    df['total_trans_count'] = df.get('total_trans_count', pd.Series(index=df.index)).fillna(0)

    df['days_since_first_transaction'] = (df['impression_timestamp'] - df['first_trans_date']).dt.days.clip(lower=0)
    df['days_since_last_transaction'] = (df['impression_timestamp'] - df['last_trans_date']).dt.days.clip(lower=0)
    df['transaction_cadence'] = df['days_since_first_transaction'] / (df['total_trans_count'] + 1e-6)


    # === 10. Additional Features (Example: Industry-level averages from offer_meta_df) ===
    print("Step 10/10: Additional features (Industry averages)...")
    required_cols_for_industry_agg = ['Card Member Industry Code', 'Discount Rate', 'Start Timestamp', 'End Timestamp']
    if all(col in df.columns for col in required_cols_for_industry_agg):
        df['offer_duration_days_meta'] = (pd.to_datetime(df['End Timestamp'], errors='coerce') - pd.to_datetime(df['Start Timestamp'], errors='coerce')).dt.days.clip(lower=0)

        industry_agg = df.groupby('Card Member Industry Code').agg(
            avg_discount_in_industry=('Discount Rate', 'mean'),
            avg_duration_in_industry=('offer_duration_days_meta', 'mean')
        ).reset_index()

        df = df.merge(industry_agg, on='Card Member Industry Code', how='left')

        df['avg_discount_in_industry'].fillna(df['Discount Rate'].mean(), inplace=True)
        df['avg_duration_in_industry'].fillna(df['offer_duration_days_meta'].mean(), inplace=True)

        df['discount_vs_industry_avg'] = df['Discount Rate'] - df['avg_discount_in_industry']
        df['duration_vs_industry_avg'] = df['offer_duration_days'] - df['avg_duration_in_industry']
        df.drop(columns=['offer_duration_days_meta'], errors='ignore', inplace=True)
    else:
        print("Warning: Required offer metadata columns not found. Skipping industry average features.")
        df['avg_discount_in_industry'] = 0
        df['avg_duration_in_industry'] = 0
        df['discount_vs_industry_avg'] = 0
        df['duration_vs_industry_avg'] = 0


    # Final cleanup
    df.replace([np.inf, -np.inf], np.nan, inplace=True)
    numerical_cols = df.select_dtypes(include=np.number).columns
    df[numerical_cols] = df[numerical_cols].fillna(0)

    print("Feature engineering complete.")
    return df

# Example usage:
# Assuming train_df, test_df, ad_event_df, transactions_df are loaded and mapped
train_df['Customer ID'] = train_df['Customer ID'].astype(str)
test_df['Customer ID'] = test_df['Customer ID'].astype(str)
ad_event_df['Customer ID'] = ad_event_df['Customer ID'].astype(str)
transactions_df['Customer ID'] = transactions_df['Customer ID'].astype(str)

# It's also a good idea to remove any accidental whitespace
train_df['Customer ID'] = train_df['Customer ID'].str.strip()
ad_event_df['Customer ID'] = ad_event_df['Customer ID'].str.strip()
transactions_df['Customer ID'] = transactions_df['Customer ID'].str.strip()


train_df = create_features(train_df, ad_event_df, transactions_df)
test_df = create_features(test_df, ad_event_df, transactions_df)
```

```python
def preprocess_final(df, is_train=True, artifacts=None):
    print("\n🚀 Starting final preprocessing...")

    # Clean column name suffixes from merges
    df.columns = df.columns.str.replace('_x', '', regex=False)
    df.columns = df.columns.str.replace('_y', '', regex=False)

    if is_train:
        artifacts = {}

        # Step 1: Drop raw columns with too much missing data (ignore engineered features)
        cols_to_ignore = ['target', 'customer_id', 'offer_id']
        engineered_prefixes = (
            'ad_', 'trans_', 'first_', 'last_', 'total_', 'unique_',
            'avg_', 'std_', 'discount_', 'f377', 'brand'
        )
        raw_cols = [
            col for col in df.columns
            if col not in cols_to_ignore and not col.startswith(engineered_prefixes)
        ]

        missing_threshold = 0.70
        missing_percentages = df[raw_cols].isnull().mean()
        cols_to_drop_missing = missing_percentages[missing_percentages > missing_threshold].index.tolist()

        if cols_to_drop_missing:
            print(f"📉 Dropping {len(cols_to_drop_missing)} raw columns with >{missing_threshold*100:.0f}% missing: {cols_to_drop_missing}")

        artifacts['cols_to_drop_missing'] = cols_to_drop_missing
    else:
        cols_to_drop_missing = artifacts.get('cols_to_drop_missing', [])


    # Step 2: Drop manually unnecessary columns (IDs, dates)
    cols_to_drop_manual = [col for col in df.columns if 'date' in col or 'time' in col]
    id_cols_to_drop = ['impression_id', 'placement_id', 'product_id']
    df.drop(columns=cols_to_drop_manual + id_cols_to_drop, inplace=True, errors='ignore')


    # Step 3: Encode categorical columns
    cols_to_exclude_from_encoding = ['id1']  # Exclude IDs here
    categorical_cols = [
    col for col in df.select_dtypes(include=['object', 'category', 'string']).columns
    if col not in cols_to_exclude_from_encoding
    ]

    if is_train:
        print(f"🔤 Encoding {len(categorical_cols)} categorical columns...")
        from sklearn.preprocessing import LabelEncoder
        label_encoders = {}
        for col in categorical_cols:
            # Handle potential 'missing' category introduced earlier
            if 'missing' in df[col].unique():
                 le = LabelEncoder()
                 # Temporarily convert to object to handle 'missing' category, then encode
                 df[col] = df[col].astype(str)
                 df[col] = le.fit_transform(df[col])
                 label_encoders[col] = le # Store the encoder
            else:
                le = LabelEncoder()
                df[col] = le.fit_transform(df[col])
                label_encoders[col] = le # Store the encoder


        artifacts['label_encoders'] = label_encoders
    else:
        print(f"🔤 Encoding {len(categorical_cols)} categorical columns using stored encoders...")
        label_encoders = artifacts.get('label_encoders', {})
        for col in categorical_cols:
            if col in label_encoders:
                le = label_encoders[col]
                # Handle unseen labels in test data during transformation
                # Add unseen labels to the encoder's classes
                for label in df[col].unique():
                    if label not in le.classes_:
                         le.classes_ = np.append(le.classes_, label) # Add the new label to the classes
                df[col] = le.transform(df[col])
            else:
                print(f"Warning: Label encoder not found for column '{col}'. Skipping encoding.")
                # Optionally, handle unseen columns in test data by dropping or another strategy
                # df.drop(columns=[col], inplace=True, errors='ignore') # Example: drop unseen columns


    # Step 4: Fill remaining missing numerical values (engineered features might have NaNs)
    numerical_cols = df.select_dtypes(include=np.number).columns.tolist()
    if is_train:
        print(f" filling numeric cols")
        imputation_values = df[numerical_cols].median() # Use median for robustness
        artifacts['imputation_values'] = imputation_values
    else:
        imputation_values = artifacts.get('imputation_values', {})


    if not imputation_values.empty:
        df[numerical_cols] = df[numerical_cols].fillna(imputation_values)
    else:
        # If no imputation values were calculated (e.g., no numerical columns in train), fill with 0
        df[numerical_cols] = df[numerical_cols].fillna(0)


    # Step 5: Separate target variable if in train mode
    target = None
    if is_train and 'y' in df.columns:
        target = df['y']
        df = df.drop(columns=['y'])
        print("Target variable 'y' separated.")

    print("✅ Final preprocessing complete.")
    if is_train:
        return df, target, artifacts
    else:
        return df, target # Return target as None for test data if not present


if train_df is not None:
    processed_train, target, artifacts = preprocess_final(train_df.copy(), is_train=True)
    processed_test, test_features = preprocess_final(test_df.copy(), is_train=False, artifacts=artifacts)

    artifacts_path = os.path.join(data_folder_path, 'ranking_artifacts.pkl')
    with open(artifacts_path, 'wb') as f: pickle.dump(artifacts, f)
    print(f"Preprocessing artifacts saved to {artifacts_path}")
```

```python
import lightgbm as lgb
from sklearn.model_selection import GroupKFold
import pandas as pd
import optuna
import numpy as np
import os
import pickle

# Assuming processed_train, target, and train_df are available from previous steps
# and that 'Customer ID' is the grouping column in train_df.

# Function to calculate Mean Average Precision at k (MAP@k)
def mapk(actual, predicted, k=7):
    """
    Calculates Mean Average Precision at k.

    Args:
        actual (list): A list of lists of ground truth items for each query.
        predicted (list): A list of lists of predicted items for each query.
        k (int): The maximum number of predicted items to consider.

    Returns:
        float: The Mean Average Precision at k.
    """
    return np.mean([apk(a, p, k) for a, p in zip(actual, predicted)])

def apk(actual, predicted, k=7):
    """
    Calculates Average Precision at k.

    Args:
        actual (list): A list of ground truth items.
        predicted (list): A list of predicted items.
        k (int): The maximum number of predicted items to consider.

    Returns:
        float: The Average Precision at k.
    """
    if not actual:
        return 0.0

    score = 0.0
    num_hits = 0.0

    for i, p in enumerate(predicted[:k]):
        if p in actual and p not in predicted[:i]:
            num_hits += 1.0
            score += num_hits / (i + 1.0)

    return score / min(len(actual), k)


if 'processed_train' in locals() and 'target' in locals() and 'train_df' in locals():
    print("\nStarting hyperparameter tuning with Optuna...")

    group_col = 'Customer ID'
    if group_col not in train_df.columns:
        print(f"Error: Grouping column '{group_col}' not found in train_df.")
    else:
        # Use GroupKFold to split the data while keeping groups (customers) together
        gkf = GroupKFold(n_splits=5)

        # Define the Optuna objective function
        def objective(trial):
            params = {
                'objective': 'lambdarank',
                'metric': 'ndcg',
                'n_estimators': trial.suggest_int('n_estimators', 1000, 6000),
                'learning_rate': trial.suggest_float('learning_rate', 0.005, 0.05),
                'num_leaves': trial.suggest_int('num_leaves', 20, 300),
                'max_depth': trial.suggest_int('max_depth', 5, 16),
                'min_child_samples': trial.suggest_int('min_child_samples', 20, 100),
                'subsample': trial.suggest_float('subsample', 0.6, 1.0),
                'colsample_bytree': trial.suggest_float('colsample_bytree', 0.5, 1.0),
                'reg_alpha': trial.suggest_float('reg_alpha', 1e-3, 10.0, log=True),
                'reg_lambda': trial.suggest_float('reg_lambda', 1e-3, 10.0, log=True),
                'seed': 42,
                'n_jobs': -1,
                'verbose': -1,
                'scale_pos_weight': trial.suggest_float('scale_pos_weight', 1.0, 20.0, log=True) # Add scale_pos_weight
            }

            # Use only the first fold for tuning for demonstration purposes
            train_idx, val_idx = next(gkf.split(processed_train, target, train_df[group_col]))

            # Use the indices to split the processed data and target
            X_train_fold, X_val_fold = processed_train.iloc[train_idx].copy(), processed_train.iloc[val_idx].copy()
            y_train_fold, y_val_fold = target.iloc[train_idx].copy(), target.iloc[val_idx].copy()

            # Ensure 'Offerings ID' and 'Card Member Industry Code' exist and are numeric/encoded
            offer_id_col = 'Offerings ID'
            industry_code_col = 'Card Member Industry Code'
            placeholder_date = pd.to_datetime('1900-01-01') # Ensure placeholder_date is defined


            # Calculate smoothed rates if the necessary columns exist
            if offer_id_col in X_train_fold.columns and industry_code_col in X_train_fold.columns:
                m = 30 # Smoothing factor
                global_offer_rate = y_train_fold.mean()

                # Calculate smoothed rates on the training fold
                offer_rate_map = X_train_fold.copy()
                offer_rate_map['target'] = y_train_fold
                offer_rate_map = offer_rate_map.groupby(offer_id_col)['target'].agg(['mean', 'count'])
                offer_rate_map['smoothed_offer_rate'] = (offer_rate_map['mean'] * offer_rate_map['count'] + global_offer_rate * m) / (offer_rate_map['count'] + m)

                industry_rate_map = X_train_fold.copy()
                industry_rate_map['target'] = y_train_fold
                industry_rate_map = industry_rate_map.groupby(industry_code_col)['target'].agg(['mean', 'count'])
                industry_rate_map['smoothed_industry_rate'] = (industry_rate_map['mean'] * industry_rate_map['count'] + y_train_fold.mean() * m) / (industry_rate_map['count'] + m)

                # Merge smoothed rates back to the folds
                X_train_fold = pd.merge(X_train_fold, offer_rate_map[['smoothed_offer_rate']], left_on=offer_id_col, right_index=True, how='left')
                X_train_fold = pd.merge(X_train_fold, industry_rate_map[['smoothed_industry_rate']], left_on=industry_code_col, right_index=True, how='left')
                X_val_fold = pd.merge(X_val_fold, offer_rate_map[['smoothed_offer_rate']], left_on=offer_id_col, right_index=True, how='left')
                X_val_fold = pd.merge(X_val_fold, industry_rate_map[['smoothed_industry_rate']], left_on=industry_code_col, right_index=True, how='left')

                # Fill any NaNs created by the merge (for categories not seen in train_fold)
                X_train_fold['smoothed_offer_rate'] = X_train_fold['smoothed_offer_rate'].fillna(global_offer_rate)
                X_val_fold['smoothed_offer_rate'] = X_val_fold['smoothed_offer_rate'].fillna(global_offer_rate)
                X_train_fold['smoothed_industry_rate'] = X_train_fold['smoothed_industry_rate'].fillna(y_train_fold.mean())
                X_val_fold['smoothed_industry_rate'] = X_val_fold['smoothed_industry_rate'].fillna(y_val_fold.mean()) # Use y_val_fold.mean() for validation set


            model = lgb.LGBMRanker(**params)

            # For LightGBM Ranker, the group information is needed
            # Get group sizes for train and validation folds
            groups_train_fold_optuna = train_df.iloc[train_idx][group_col]
            groups_val_fold_optuna = train_df.iloc[val_idx][group_col]

            train_group_sizes_optuna = groups_train_fold_optuna.value_counts().sort_index().to_numpy()
            val_group_sizes_optuna = groups_val_fold_optuna.value_counts().sort_index().to_numpy()


            model.fit(X_train_fold, y_train_fold,
                      group=train_group_sizes_optuna,
                      eval_set=[(X_val_fold, y_val_fold)],
                      eval_group=[val_group_sizes_optuna],
                      eval_metric='ndcg',
                      callbacks=[lgb.early_stopping(100, verbose=False)])

            # Make predictions on the validation set
            val_preds = model.predict(X_val_fold)

            # Create a DataFrame for evaluation
            val_df = pd.DataFrame({
                'customer_id': groups_val_fold_optuna, # Use the unsorted groups for correct alignment
                'offer_id': train_df.loc[val_idx, 'Offerings ID'], # Use original train_df for Offerings ID and unsorted index
                'target': y_val_fold,
                'pred_score': val_preds
                })

            # Sort by customer and prediction score for ranking
            val_df = val_df.sort_values(['customer_id', 'pred_score'], ascending=[True, False])

            # Get ranked lists of offer_ids for each customer based on predictions
            predicted_lists = val_df.groupby('customer_id')['offer_id'].apply(list)

            # Get the actual lists of offered_ids where target is 1 for each customer
            actual_lists = val_df[val_df['target'] == 1].groupby('customer_id')['offer_id'].apply(list)

            # Align actual lists with predicted lists based on customer_id index
            aligned_actual = [actual_lists.get(user_id, []) for user_id in predicted_lists.index]

            # Calculate MAP@k
            score = mapk(aligned_actual, predicted_lists.tolist(), k=7)

            return score


        # Run Optuna study
        study = optuna.create_study(direction='maximize')
        study.optimize(objective, n_trials=20, timeout=1800) # Adjust n_trials and timeout as needed

        print("Number of finished trials: ", len(study.trials))
        print("Best trial:")
        trial = study.best_trial
        print("  Value: ", trial.value)
        print("  Params: ")
        best_params = trial.params
        for key, value in best_params.items():
            print(f"    {key}: {value}")

        # Save the best parameters
        best_params_path = os.path.join(data_folder_path, 'best_params.pkl')
        with open(best_params_path, 'wb') as f:
            pickle.dump(best_params, f)
        print(f"Best parameters saved to {best_params_path}")


else:
    print("Processed data or target not found. Please run the preprocessing step first.")
```
